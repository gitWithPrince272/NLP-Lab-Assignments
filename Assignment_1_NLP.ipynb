```python
# Function to clean text using all 7 cleaning steps

def clean_text(text):
    # 1. Lowercase
    text = text.lower()

    # 2. Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    # 3. Expand contractions
    for contraction, expanded in contractions.items():
        text = text.replace(contraction, expanded)

    # 4. Remove special characters
    text = re.sub(r'[^a-zA-Z0-9\s.,!?]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # 5. Remove excessive repeated letters
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # 6. Remove URLs
    text = re.sub(r'www\.\S+|https?://\S+', '', text)

    # 7. Remove punctuation
    text = text.translate(
        str.maketrans('', '', string.punctuation)
    )

    # Final whitespace cleanup
    text = re.sub(r'\s+', ' ', text).strip()

    return text

print("Cleaning function created successfully!")
```

```python
# Task 2 - Experiment 1
# Load and tokenize Brown Corpus

import nltk

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import brown
from nltk.tokenize import word_tokenize

# Select the news category
brown_text = " ".join(brown.words(categories="news"))

print("Brown Corpus loaded successfully!")
print("Number of characters:", len(brown_text))
print("\nFirst 500 characters:")
print(brown_text[:500])
```

```python
# Tokenize the Brown Corpus text

tokens = word_tokenize(brown_text)

print("Tokenization completed successfully!")
print("Total tokens:", len(tokens))
print("\nFirst 30 tokens:")
print(tokens[:30])
```

```python
# Tokenize the Brown Corpus

tokens = word_tokenize(brown_text)

print("Tokenization completed successfully!")
print("Total number of tokens:", len(tokens))

print("\nFirst 30 tokens:")
print(tokens[:30])
```

```python
from collections import Counter

# Calculate raw word frequencies
word_frequency = Counter(tokens)

print("Raw word frequency calculated successfully!")

print("\nTop 20 most frequent words:")
for word, count in word_frequency.most_common(20):
    print(f"{word}: {count}")
```

```python
import matplotlib.pyplot as plt

# Get top 20 most frequent words
top_20 = word_frequency.most_common(20)

words = [word for word, count in top_20]
frequencies = [count for word, count in top_20]

plt.figure(figsize=(12, 6))
plt.bar(words, frequencies)

plt.title("Top 20 Most Frequent Words - Brown Corpus")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Experiment 3: Porter Stemming

from nltk.stem import PorterStemmer
from collections import Counter

# Create Porter Stemmer
stemmer = PorterStemmer()

# Apply stemming to all tokens
stemmed_tokens = [stemmer.stem(word) for word in tokens]

# Calculate frequency of stemmed words
stemmed_frequency = Counter(stemmed_tokens)

print("Porter Stemming completed successfully!")

print("\nTop 20 stemmed words with frequency:")
for word, count in stemmed_frequency.most_common(20):
    print(f"{word}: {count}")
```

```python
# Experiment 4: WordNet Lemmatization

import nltk
from nltk.stem import WordNetLemmatizer
from collections import Counter

# Download required WordNet resources
nltk.download('wordnet')
nltk.download('omw-1.4')

# Create lemmatizer
lemmatizer = WordNetLemmatizer()

# Apply lemmatization
lemmatized_tokens = [
    lemmatizer.lemmatize(word)
    for word in tokens
]

# Calculate frequency
lemmatized_frequency = Counter(lemmatized_tokens)

print("WordNet Lemmatization completed successfully!")

print("\nTop 20 lemmatized words with frequency:")
for word, count in lemmatized_frequency.most_common(20):
    print(f"{word}: {count}")
```

```python
# Experiment 5: Stemming vs Lemmatization Comparison

# Select 15 common words from the Brown Corpus
common_words = [
    "running", "runs", "played", "playing", "studies",
    "studying", "better", "children", "mice", "cars",
    "houses", "quickly", "easily", "government", "making"
]

comparison = []

for word in common_words:
    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word)

    comparison.append({
        "Original Word": word,
        "Stemmed": stemmed,
        "Lemmatized": lemmatized
    })

comparison_df = pd.DataFrame(comparison)

display(comparison_df)
```

```python
# Compare the length of stemmed and lemmatized words

comparison_plot = comparison_df.copy()

comparison_plot["Stemmed Length"] = comparison_plot["Stemmed"].str.len()
comparison_plot["Lemmatized Length"] = comparison_plot["Lemmatized"].str.len()

x = np.arange(len(comparison_plot))
width = 0.35

plt.figure(figsize=(14, 6))

plt.bar(
    x - width/2,
    comparison_plot["Stemmed Length"],
    width,
    label="Stemmed"
)

plt.bar(
    x + width/2,
    comparison_plot["Lemmatized Length"],
    width,
    label="Lemmatized"
)

plt.xticks(x, comparison_plot["Original Word"], rotation=45)
plt.xlabel("Original Words")
plt.ylabel("Number of Characters")
plt.title("Stemming vs Lemmatization")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Video File Handling

import cv2
import numpy as np
from IPython.display import Video

# Video settings
width, height = 640, 360
fps = 20
duration = 3

video_path = "sample_video.mp4"

# Create video writer
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(video_path, fourcc, fps, (width, height))

# Create frames
for i in range(fps * duration):
    frame = np.ones((height, width, 3), dtype=np.uint8) * 255

    # Moving circle
    x = int((i / (fps * duration - 1)) * (width - 100)) + 50
    y = height // 2

    cv2.circle(frame, (x, y), 40, (255, 0, 0), -1)

    out.write(frame)

out.release()

print("Video file created successfully!")

# Display the video
Video(video_path, embed=True)
```
